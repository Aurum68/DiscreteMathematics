```python
from main import nx, edges, x, y, matching_edges, max_matching_set
```

```python
graph = nx.Graph()
graph.add_nodes_from(list(x), bipartite=0)
graph.add_nodes_from(list(y), bipartite=1)
graph.add_edges_from(edges, color='black')
#nx.draw(graph, pos=nx.bipartite_layout(graph, x), with_labels=True)

for edge in matching_edges:
    if (edge[0] + 1, edge[1] + 1) in graph.edges:
        graph[edge[0] + 1][edge[1] + 1]['color'] = 'blue'
    elif (edge[1] + 1, edge[0] + 1) in graph.edges:
        graph[edge[1] + 1][edge[0] + 1]['color'] = 'blue'

print("Ford-Fulkerson")
edge_color_list = [graph[edge[0]][edge[1]]['color'] for edge in graph.edges]
nx.draw(graph, edge_color=edge_color_list, pos=nx.bipartite_layout(graph, x), with_labels=True)
```

```python
for edge in graph.edges:
    graph[edge[0]][edge[1]]['color'] = 'black'

for edge in max_matching_set:
    u = edge.u
    v = edge.v
    if (u + 2, v + 2) in graph.edges:
        graph[u + 2][v + 2]['color'] = 'blue'
    elif (v + 2, u + 2) in graph.edges:
        graph[v + 2][u + 2]['color'] = 'blue'

print("Khun")
edge_color_list = [graph[edge[0]][edge[1]]['color'] for edge in graph.edges]
nx.draw(graph, edge_color=edge_color_list, pos=nx.bipartite_layout(graph, x), with_labels=True)
```
